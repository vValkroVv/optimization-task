```python
import pandas as pd
from tbparse import SummaryReader
from collections import defaultdict, OrderedDict
from tqdm.auto import tqdm
import wandb
import yaml

HOST = 
KEY  = 

wandb.login(relogin=True, host=HOST, key=KEY)
```

```python
# names of experiments (i.e. folders in wandb)
exp_names = [
    'gigar3_180ckpt_contin',
    'gigar3_180ckpt_contin',
]

# full paths to tensorboard logs
log_dirs = [
    '/home/jovyan/evgenijkkk/tensorboard_180ckpt/gigar3_180ckpt_contin/filtered_encc',
    '/home/jovyan/evgenijkkk/tensorboard_180ckpt/gigar3_180ckpt_contin/fineweb_edu_frw_proportional_mix',
]

# names of runs in wandb
run_names = [
    'filtered_encc',
    'fineweb_edu_frw_proportional_mix',
]

# names of yaml config files
log_files = [
    '180ckpt_filtered_encc.yaml',
    '180ckpt_fineweb_edu_frw_proportional_mix.yaml',
]
```

```python
for log_dir, run_name, log_file, exp_name in zip(log_dirs, run_names, log_files, exp_names):
    print(f'Started {log_dir} || {run_name} || {log_file} || {exp_name}')
    
    reader = SummaryReader(log_dir)
    df = reader.scalars
    print(df)
    
    run_logs = defaultdict(dict)
    for _, row in df.iterrows():
        run_logs[row.step].update({
            row.tag: row.value
        })
    run_logs = OrderedDict(sorted(run_logs.items()))
    
    with open(log_file) as f:
        cfg = yaml.safe_load(f)
        
    run = wandb.init(
        name=run_name,
        project=exp_name,
        config=cfg
    )

    for step, data in tqdm(run_logs.items()):
        wandb.log(data=data, step=step)

    wandb.finish()
    #run.log_code()
```
